##  Objective

#### This section imports NumPy for working with the datasets as matrices and Matplotlib for creating the required graphs and plots.

## 6. BMI calculation for female dataset

#### BMI is calculated for each female participant from weight and height, with height first converted from centimetres to metres. The calculated BMI values are then added as the eighth column of the female matrix.

```python
height_m = female[:, 1] / 100

bmi = female[:, 0] / (height_m ** 2)

female = np.column_stack((female, bmi))

print(female.shape)
```

### Interpretation

#### the BMI values have been calculated for all female participants and added as the 8th column. The female matrix now has 4221 rows and 8 columns, with BMI added to the original  body measurements.

## 7. Standardising the Female Measurements

#### A new matrix called zfemale is created by converting all eight columns of the female matrix into z-scores. This places the different measurements on a common standardised scale.

```python
zfemale = (female - np.mean(female, axis=0)) / np.std(female, axis=0)

print(zfemale[:5])
```

### Interpretation

#### A new matrix called zfemale has been created from the female dataset. It contains the standardised values for all 8 columns, including the BMI column added earlier. The values are now on a common z-score scale, which makes the different measurements easier to compare.Values above 0 are above the average, while values below 0 are below the average.

## 8. Scatterplot Matrix and Correlation Analysis

#### The standardised height, weight, waist circumference, hip circumference and BMI values are selected from zfemale. A scatterplot matrix is used to examine their relationships visually, followed by Pearson and Spearman correlation coefficients to measure the strength and direction of the relationships.

```python
data = zfemale[:, [1, 0, 6, 5, 7]]
```

```python
plt.figure(figsize=(10, 10))

for i in range(5):
    for j in range(5):
        plt.subplot(5, 5, i * 5 + j + 1)
        plt.scatter(data[:, j], data[:, i])
        plt.xlabel(["Height", "Weight", "Waist", "Hip", "BMI"][j])
        plt.ylabel(["Height", "Weight", "Waist", "Hip", "BMI"][i])

plt.tight_layout()
plt.show()
```

### Interpretation

#### The plots show a clear positive relationship between weight, waist, hip and BMI. The points for these variables follow a noticeable upward pattern. Height is less closely related, especially to BMI, where the points are more spread out.

###  Correlation analysis for standardised female variables

```python
variable_names = ["Height", "Weight", "Waist", "Hip", "BMI"]

# Pearson correlation matrix
pearson_corr = np.corrcoef(data, rowvar=False)

print("Pearson correlation matrix:")
print(pearson_corr)


# Spearman correlation matrix
ranked_data = np.zeros_like(data, dtype=float)

for i in range(data.shape[1]):
    ranked_data[:, i] = np.argsort(np.argsort(data[:, i])) + 1

spearman_corr = np.corrcoef(ranked_data, rowvar=False)

print("\nSpearman correlation matrix:")
print(spearman_corr)
```

```python
print("Variables:", variable_names)

print("\nPearson correlations:")
for i in range(len(variable_names)):
    for j in range(i + 1, len(variable_names)):
        print(
            f"{variable_names[i]} vs {variable_names[j]}: "
            f"{pearson_corr[i, j]:.3f}"
        )

print("\nSpearman correlations:")
for i in range(len(variable_names)):
    for j in range(i + 1, len(variable_names)):
        print(
            f"{variable_names[i]} vs {variable_names[j]}: "
            f"{spearman_corr[i, j]:.3f}"
        )
```

### Interpretation

#### The results show that weight, waist circumference, hip circumference and BMI have very strong positive relationships with each other. The strongest relationships are between weight and hip (0.947), weight and BMI (0.946), and hip and BMI (0.944). Weight also has a strong relationship with waist (0.905).

#### Height has weaker relationships with the other variables. Its strongest relationship is with weight (0.345), while its relationships with waist (0.127), hip (0.203) and BMI (0.033) are weak.

#### The Pearson and Spearman coefficients are very similar for every pair, indicating that the relationships observed are consistent when measured using both linear correlation and rank-based correlation.

## 11. Advantages and Disadvantages of BMI, Waist-to-Height Ratio and Waist-to-Hip Ratio

#### BMI, waist-to-height ratio and waist-to-hip ratio are discussed in terms of their specific strengths and limitations as measures of body composition and fat distribution.

## BMI
### Advantages 
    1. Simple to calculate and easy to understand
    2. Useful for getting a general idea of whether a person's weight is appropirate to their   height.
    3. Helps to campare across large group of people
    4. People can use this BMI index to improve their health by checking on their calorie intake.

### Disadvantages
    1. Does not distinguish between muscle and body fat
    2. Does not show exactly at what part of the body the person has accumulated fat.
    3. Two people with the same BMI can have very different body compositions.

## Waist to height Ratio

### Advantages
    1. Takes waist size and height into account.Relates waist circumference directly to height and indicates central body size.
    2. Gives an indication of central body and abdominal fat distribution.
    3. It can easily distinguish between people of similar BMI who has different abdominal fat.
    4. Easy to calculate and interpret.

### Disadvantages
    1. Difference on waist measurement can affect the ratio.
    2. It doesnt measure total body fat.
    3. A single ratio cannot determine the ful body fat distribution.

## Waist-to-Hip Ratio

### Advantages 
    1. Compares waist size to hip size.
    2. Useful for distinguishing different body-fat distribution patterns, even when people have similar overall body size.
    3. Uses two simple circumference measurements and is therefore easy to calculate from the dataset.

### Disadvantages
    1. The value can be influenced by hip size and body shape, not just abdominal fat.
    2. It does not directly measure total body fat or body composition.
    3. Both waist and hip measurements must be taken consistently , measurement differences can change the ratio.

## 12. Standardised Measurements of Participants with Lowest and Highest BMI

#### The female participants are ranked according to BMI using np.argsort(). The five participants with the lowest BMI and the five with the highest BMI are then selected, and their standardised body measurements are examined to identify differences between the two groups.

```python
bmi_order = np.argsort(female[:, 7])

lowest_bmi = bmi_order[:5]
highest_bmi = bmi_order[-5:]

print("5 lowest BMI:")
print(zfemale[lowest_bmi])

print("\n5 highest BMI:")
print(zfemale[highest_bmi])
```

### Interpretation

#### The five participants with the lowest BMI have measurements that are generally below the average. The five participants with the highest BMI show the opposite pattern, with higher weight and waist and hip measurements. Overall, the results show a clear difference in body measurements between the lowest- and highest-BMI groups.
